Q7.

1. Duplicate Check

Verify that there are no duplicate records, especially in the Order_ID column, to avoid primary key violations.

2. Missing Value Check

Identify and handle NULL values in important columns like Sales_Amount to prevent incomplete data and constraint issues.

3. Data Type Validation

Ensure that all columns contain correct data types (e.g., Sales_Amount should be numeric, not text).

4. Date Format Standardization

Check that all values in Order_Date follow a consistent format (e.g., YYYY-MM-DD) for proper parsing.

5. Schema Validation

Confirm that the dataset structure (column names, order, and data types) matches the target database table.

6. Constraint Validation

Ensure all constraints are satisfied:

   Primary Key (unique Order_ID)

   NOT NULL (mandatory fields are filled)

7. Data Consistency Check

Verify that values are logically correct and standardized (e.g., no mixed formats or inconsistent entries)
